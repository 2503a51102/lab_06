```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

from google.colab import files

uploaded = files.upload()
file_name = next(name for name in uploaded if name.endswith(".csv"))

df = pd.read_csv(file_name)

print(df.shape)
print(df.columns.tolist())
display(df.head())
```

```python
fig1 = px.histogram(
    df,
    x="GPA",
    nbins=25,
    marginal="box",
    title="Distribution of Student GPA",
    hover_data=["Student_ID", "Current_Semester_Performance"]
)

fig1.update_layout(
    xaxis_title="GPA",
    yaxis_title="Number of Students",
    template="plotly_white"
)

fig1.show()
```

```python
fig2 = px.scatter(
    df,
    x="Attendance_Rate",
    y="GPA",
    color="Dropout_Risk",
    hover_data=[
        "Student_ID",
        "Study_Hours_per_Week"
    ],
    title="Attendance vs GPA by Dropout Risk",
    opacity=0.7
)

fig2.update_layout(
    xaxis_title="Attendance Rate",
    yaxis_title="GPA",
    template="plotly_white"
)

fig2.show()
```

```python
fig3 = px.box(
    df,
    x="Current_Semester_Performance",
    y="GPA",
    color="Current_Semester_Performance",
    points="all",
    hover_data=["Student_ID", "Attendance_Rate"],
    title="GPA by Performance Category"
)

fig3.update_layout(
    xaxis_title="Performance Category",
    yaxis_title="GPA",
    showlegend=False,
    template="plotly_white"
)

fig3.show()
```

```python
fig4 = px.scatter_3d(
    df,
    x="Study_Hours_per_Week",
    y="Attendance_Rate",
    z="GPA",
    color="Current_Semester_Performance",
    hover_data=[
        "Student_ID",
        "Dropout_Risk"
    ],
    title="3D Student Performance Analysis",
    opacity=0.7
)

fig4.update_layout(template="plotly_white")

fig4.show()
```

```python
animation_df = df.copy()

x_min = animation_df["Attendance_Rate"].min()
x_max = animation_df["Attendance_Rate"].max()
y_min = animation_df["GPA"].min()
y_max = animation_df["GPA"].max()

fig5 = px.scatter(
    animation_df,
    x="Attendance_Rate",
    y="GPA",
    size="Credits_Completed",
    size_max=12,
    color="Current_Semester_Performance",
    hover_name="Student_ID",
    hover_data=["Study_Hours_per_Week", "Dropout_Risk"],
    range_x=[x_min - 2, x_max + 2],
    range_y=[y_min - 0.5, y_max + 0.5],
    title="Student Performance Analysis: Attendance vs GPA"
)

fig5.update_layout(template="plotly_white")

fig5.show()
```

```python
metrics = {
    "Attendance Rate": df["Attendance_Rate"],
    "GPA": df["GPA"],
    "Online Learning Hours": df["Online_Learning_Hours"],
    "Library Usage Hours": df["Library_Usage_Hours"]
}

fig6 = go.Figure()

for i, (metric_name, values) in enumerate(metrics.items()):
    fig6.add_trace(
        go.Scatter(
            x=df["Study_Hours_per_Week"],
            y=values,
            mode="markers",
            name=metric_name,
            visible=(i == 0),
            text=df["Student_ID"],
            marker=dict(size=7, opacity=0.6),
            hovertemplate=(
                "Student ID: %{text}<br>"
                "Study Hours: %{x}<br>"
                + metric_name + ": %{y}<extra></extra>"
            )
        )
    )

buttons = []

for i, metric_name in enumerate(metrics.keys()):
    buttons.append(
        dict(
            label=metric_name,
            method="update",
            args=[
                {"visible": [j == i for j in range(len(metrics))]},
                {
                    "title": "Study Hours vs " + metric_name,
                    "yaxis": {"title": metric_name}
                }
            ]
        )
    )

fig6.update_layout(
    title="Study Hours vs Attendance Rate",
    xaxis_title="Study Hours per Week",
    yaxis_title="Attendance Rate",
    updatemenus=[
        dict(
            buttons=buttons,
            direction="down",
            x=1.02,
            y=1
        )
    ],
    margin=dict(r=220),
    template="plotly_white"
)

fig6.show()
```

```python
fig = px.bar(
    df["Current_Semester_Performance"].value_counts().reset_index(),
    x="Current_Semester_Performance",
    y="count",
    title="Students by Performance Category",
    text="count"
)

fig.show()
```

```python
fig = px.pie(
    df,
    names="Dropout_Risk",
    title="Dropout Risk Distribution",
    hole=0.3
)

fig.show()
```

```python
performance_gpa = (
    df.groupby("Past_Academic_Performance", as_index=False)["GPA"]
    .mean()
    .sort_values("GPA")
)

fig = px.line(
    performance_gpa,
    x="Past_Academic_Performance",
    y="GPA",
    markers=True,
    title="Average GPA by Past Academic Performance"
)

fig.show()
```
